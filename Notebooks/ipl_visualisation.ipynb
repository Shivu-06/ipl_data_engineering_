```python
import pandas as pd
import matplotlib.pyplot as plt

matches = pd.read_csv("../cleaned_data/clean_matches.csv")

top_teams = matches['winner'].value_counts().head(10)

plt.figure(figsize=(10,5))
top_teams.plot(kind='bar')

plt.title("Top IPL Teams by Wins")
plt.xlabel("Team")
plt.ylabel("Wins")

plt.show()
```
